```python
#week 4
import pandas as pd
import numpy as np
from scipy import stats
df=pd.read_csv("auto-mpg.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
dfclean=df.dropna()
```

```python
lower=dfclean['mpg'].quantile(0.05)
upper=dfclean['mpg'].quantile(0.95)
dfp=dfclean[(dfclean["mpg"]>=lower) & (dfclean["mpg"]<=upper)]
dfp
```

```python
mean=dfclean["mpg"].mean()
std=dfclean["mpg"].std()
dfstd=dfclean[(dfclean["mpg"]>=mean-2*std) & (dfclean["mpg"]<=mean+3*std)]
dfstd
```

```python
q1=dfclean["mpg"].quantile(0.25)
q3=dfclean["mpg"].quantile(0.75)
iqr=q3-q1
dfiq=dfclean[(dfclean["mpg"]>=q1-1.5*iqr) & (dfclean["mpg"]<=q3+1.5*iqr)]
dfiq
```

```python
z=np.abs(stats.zscore(dfclean["mpg"]))
dfzscore=dfclean[z<3]
dfzscore
```

```python
lower_cap = df["mpg"].quantile(0.05)
upper_cap = df["mpg"].quantile(0.95)

df["mpg"] = df["mpg"].clip(lower_cap, upper_cap)
```

```python
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].mean())
```

```python
#Feature Extraction PCA LDA SVD
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris(as_frame=True)
df=iris.frame
df.head()
```

```python
X=df.iloc[:,:-1]
Y=df.iloc[:,-1]
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)
print(X_pca[:5])
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda=LinearDiscriminantAnalysis(n_components=2)
X_lda=lda.fit_transform(X_scaled,Y)
print(X_lda[:5])
```

```python
from sklearn.decomposition import TruncatedSVD
svd=TruncatedSVD(n_components=2)
X_svd=svd.fit_transform(X_scaled)
print(X_svd[:5])
```
